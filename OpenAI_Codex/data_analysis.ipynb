# Titanic Missing Values

This notebook inspects missing values in the Titanic train and test datasets.

```python
import pandas as pd

train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')

def missing_summary(df, name):
    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)
    print(f'\n{name} missing values:')
    display(missing.to_frame('missing_count'))

missing_summary(train, 'train')
missing_summary(test, 'test')
```

```python
def impute_median_numeric(df):
    df = df.copy()
    numeric_cols = df.select_dtypes(include='number').columns
    for col in numeric_cols:
        if df[col].isna().any():
            df[col] = df[col].fillna(df[col].median())
    return df

train_imputed = impute_median_numeric(train)
test_imputed = impute_median_numeric(test)

print('Train missing after numeric median impute:', int(train_imputed.isna().sum().sum()))
print('Test missing after numeric median impute:', int(test_imputed.isna().sum().sum()))
```

```python
# Create a Cabin_deck feature (first letter) and preserve missingness as its own category.
# This avoids guessing a cabin value and lets models learn from the fact that it was not recorded.
train['Cabin_deck'] = train['Cabin'].str[0].fillna('Missing')
test['Cabin_deck'] = test['Cabin'].str[0].fillna('Missing')

print(train['Cabin_deck'].value_counts().head())
```

```python
missing_summary(train_imputed, 'train')
missing_summary(test_imputed, 'test')
```
